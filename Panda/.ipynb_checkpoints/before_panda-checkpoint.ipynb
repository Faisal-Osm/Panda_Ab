# Find Max ,Min And Average Ratings For Bollywood 

```python
ratings = []
for row in data:
    if row[3] != "NULL" and row[1] == "Bollywood":
        imbd_rating = float(row[3])
        ratings.append(imbd_rating)
len(ratings)
```

```python
print(f"Maximum Rating Bollywood movies  is {max(ratings)}")
print(f"Minimum Rating Bollywood movies  is {min(ratings)}")
print(f"Average Rating Bollywood movies  is {round(sum(ratings)/len(ratings),2)}")
```

# Function

```python
def calculate_rating_status(data, industry = None):
    ratings = []
    for row in data:
        if row[3] != "NULL" and (not industry or row[1]==industry):
            ratings.append(float(row[3]))
    max_rating = max(ratings)
    min_rating = min(ratings)
    Avg_rating = round(sum(ratings)/ len(ratings),2)

    return max_rating,min_rating, Avg_rating
    
```

```python
max_imbd, min_imbd,average_imbd = calculate_rating_status(data, "Hollywood")
print(f"Maximum Rating Hollywood movies  is {max_imbd}")
print(f"Minimum Rating Hollywood movies  is {min_imbd}")
print(f"Average Rating Hollywood movies  is {average_imbd}")
```

```python
max_imbd, min_imbd,average_imbd = calculate_rating_status(data, "Bollywood")
print(f"Maximum Rating Bollywood movies  is {max_imbd}")
print(f"Minimum Rating Bollywood movies  is {min_imbd}")
print(f"Average Rating Bollywood movies  is {average_imbd}")
```

```python
max_imbd, min_imbd,average_imbd = calculate_rating_status(data)
print(f"Maximum AlllRating  movies  is {max_imbd}")
print(f"Minimum AllRating  movies  is {min_imbd}")
print(f"Average  All Rating  movies  is {average_imbd}")
```
